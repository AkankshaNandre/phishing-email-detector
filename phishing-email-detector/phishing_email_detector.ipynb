```python
print("📧 PHISHING EMAIL DETECTOR 📧")

# User enters email text
email = input("\nPaste the email message:\n")

# Suspicious keywords
phishing_keywords = [
    "urgent",
    "verify",
    "click here",
    "password",
    "bank",
    "login",
    "account suspended",
    "win money",
    "free",
    "limited time",
]

score = 0

# Convert email to lowercase
email_lower = email.lower()

# Check suspicious keywords
for keyword in phishing_keywords:
    if keyword in email_lower:
        print(f"Suspicious keyword detected: {keyword} 🚨")
        score += 1

# Check for links
if "http://" in email_lower or "https://" in email_lower:
    print("Suspicious link detected 🚨")
    score += 1

# Final score
print("\nRisk Score:", score)

# Final result
if score >= 5:
    print("⚠️ High Risk: Possible Phishing Email")
elif score >= 3:
    print("⚠️ Medium Risk: Be Careful")
else:
    print("✅ Low Risk: Email Looks Safer")
```
